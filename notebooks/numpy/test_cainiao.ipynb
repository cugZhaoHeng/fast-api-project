# 从菜鸟教程上面练习 Python 库的 Numpy 的使用

```python
import numpy as np

a = np.array([1,2,3,4,5])
print(a)
```

```python
import numpy as np

a = np.array([[1,2],[3,4]])
print(a)
```

```python
import numpy as np

a = np.arange(10, dtype=float)
print(a)
print(f'a.shape: {a.shape}')
print(f'a.ndim: {a.ndim}')
print(f'a[1]: {a[1]: .3f}')
```

```python
import numpy as np

a = np.arange(start=1, stop=12, step=2, dtype=float).reshape(2,3)
print(a)
print(f'a.shape:{a.shape}')
print(f'a.ndim: {a.ndim}')
```
